```python
import pandas as pd
```

```python
std1 = pd.read_csv("datasets_ta/esai_siswa_1.csv")
std2 = pd.read_csv("datasets_ta/esai_siswa_2.csv")
```

```python
std1.info()
std2.info()

std1.head()
std2.head()
```

```python
# Create new dataframes with only the necessary columns
df1 = std1[['Masukan Essaynya Disini guys']].rename(columns={'Masukan Essaynya Disini guys': 'Essay'})
df2 = std2[['Buatlah Esai Anda tentang "Pengalaman Selama Libur Idul Fitri dalam Memanfaatkan Teknologi"']].rename(columns={'Buatlah Esai Anda tentang "Pengalaman Selama Libur Idul Fitri dalam Memanfaatkan Teknologi"': 'Essay'})

# Combine the dataframes
combined_df = pd.concat([df1, df2], ignore_index=True)

# Save to CSV
combined_df.to_csv('esai_siswa.csv', index=False)
```
